# King County – Refactored

The cleaning and feature engineering from `King-County.ipynb` now live in the `kc_housing` package (`src/kc_housing/`). This notebook just calls the pipeline and keeps the modeling step.

```python
from sklearn.model_selection import train_test_split

from kc_housing import load_data, preprocess
from kc_housing.config import RANDOM_STATE, TEST_SIZE
from kc_housing.modeling import split_features_target, train
```

## Data cleaning + feature engineering (one call)

```python
kc_data = preprocess(load_data())
kc_data.head()
```

```python
kc_data.describe().round(2)
```

## Modeling

```python
X, y = split_features_target(kc_data)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE
)
search = train(X_train, y_train)
search.best_params_
```

```python
print("Test R^2:", round(search.best_estimator_.score(X_test, y_test), 3))
```
